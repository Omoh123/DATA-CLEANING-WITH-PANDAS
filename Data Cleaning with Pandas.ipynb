DATA CLEANING

```python
# Remove leading/trailing spaces from all column names
employees_info.columns = employees_info.columns.str.strip()
employees_info
#check column names
print(employees_info.columns)
```

```python
#finding missing values
employees_info.isnull()
#dropping empty and invalid columns(notes and join date)
employees_info.drop('JoinDate', axis=1, inplace=True)
employees_info
```

```python
#standardising country names
employees_info['Country'].replace({
    'USA': 'United States',
    'US': 'United States',
    'United States': 'United States',
    'UK': 'United Kingdom',
    'England': 'United Kingdom',
    'United Kingdom': 'United Kingdom'
}, inplace=True)
employees_info
```

```python
#convert the Salary and Age columns to numeric values
employees_info['Salary'] = pd.to_numeric(employees_info['Salary'], errors='coerce')
employees_info['Age'] = pd.to_numeric(employees_info['Age'], errors='coerce')
employees_info
```

```python
#calculate the mean salary
avg_salary = employees_info['Salary'].mean()
avg_salary
#filling the missing values in the Salary column with the mean salary
employees_info.fillna({'Salary': avg_salary}, inplace=True)
employees_info
```

```python
#calculate the average age
avg_age = employees_info['Age'].mean()
avg_age
#filling the missing values in the Age column with the mean age
employees_info.fillna({'Age': avg_age}, inplace=True)
employees_info
```

```python
#clean the email addresses
# Fix double '@' and replace 'company.com' with 'gmail.com'
employees_info['Email'] = (
    employees_info['Email']
    .str.replace(r'@@', '@', regex=True)
    .str.replace(r'company\.com', 'gmail.com', regex=True)
    .str.replace(r'mail\.com', 'gmail.com', regex=True)
    .str.replace('company', 'gmail.com', regex=False)
    # Replace 'ggmail' or 'gggmail' with 'gmail'
    .str.replace(r'g+mail', 'gmail', regex=True)
)
employees_info
```

```python
#filling the missing value in the name column with 'Unknown'
employees_info.fillna({'Name': 'Unknown'}, inplace=True)
#setting the index to be the name column
employees_info.set_index('Name', inplace=True)
employees_info
```

```python
#back up the final cleaned data
employees_info.to_csv('PRACTICE DATASETS/cleaned_employees_info.csv', index=False)
```
