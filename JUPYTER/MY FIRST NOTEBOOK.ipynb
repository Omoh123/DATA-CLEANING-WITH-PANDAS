PANDAS SERIES

```python
#creating a pandas series
g7_pop = pd.Series([35.467, 63.951, 80.94, 60.665, 127.061, 64.511, 318.523])
g7_pop
#giving it a name
g7_pop.name = "G7 population in millions"
g7_pop
#indexing
g7_pop.index = ['Canada', 'France', 'Germany', 'Italy', 'Japan', 'United Kingdom', 'United States']
g7_pop
```

```python
#summary statistics
g7_pop.describe()
```

```python
#This can be useful for identifying outliers or values that are not close to the average.
g7_pop[(g7_pop >g7_pop.mean() - g7_pop.std() / 2) | (g7_pop > g7_pop.mean() + g7_pop.std() / 2)]
```

```python
#masking
#g7_pop <=60
g7_pop[g7_pop <=60]
```

```python
#create a data frame showing the atudents scores
exam_score = pd.DataFrame({
    'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eva', 'Frank', 'Grace', 'Helen', 'Ian', 'Julia'],
    'Age': [23, 35, 45, 28, 32, 40, 29, 31, 38, 27],
    'Country': ['USA', 'Canada', 'UK', 'USA', 'Germany', 'France', 'Canada', 'UK', 'Germany', 'France'],
    'Score': [88, 92, 79, 85, 90, 87, 95, 80, 84, 91],
    'test1': [18, 17, 15, 19, 16, 20, 14, 13, 18, 17],
    'test2': [16, 19, 18, 17, 15, 20, 13, 14, 19, 16]
})
exam_score
#Add index names to the dataframe using the 'Name' column
exam_score.index = exam_score['Name']
exam_score
#drop the name column
exam_score = exam_score.drop('Name', axis=1)
exam_score
```

```python
exam_score['Average'] = (exam_score['Score'] * 0.6) + (exam_score['test1'] * 0.2) + (exam_score['test2'] * 0.2)
exam_score
```
